```python
import gymnasium as gym
from gymnasium.envs.toy_text.frozen_lake import generate_random_map
import random
from sklearn.preprocessing import StandardScaler
from collections import deque
```

```python
env = gym.make('FrozenLake-v1', render_mode="none", desc=generate_random_map(size=8), is_slippery=False)
```

```python
import random
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input
from tensorflow.keras.optimizers import Adam
import numpy as np
import tensorflow as tf

STATE_SIZE = env.observation_space.n
ACTION_SIZE = env.action_space.n
EPISODES = 1000
MEMORY_LENGTH = 5000
BATCH_SIZE = 64
STEPS = 1000
GAMMA = .95
EPSILON_DECAY = .995
EPSILON_MIN = .01
EPSILON = 1

class DQNAgent():
    def __init__(self, epsilon, epsilon_min, epsilon_decay, gamma, batch_size):
        self.epsilon = epsilon
        self.epsilon_min = epsilon_min
        self.epsilon_decay = epsilon_decay
        self.gamma = gamma
        self.batch_size = batch_size
        self.create_model()
        self.memory = deque(maxlen=MEMORY_LENGTH)
#         self.scaler = self.get_scaler()
    
    def create_model(self):
        self.model = Sequential()
        self.model.add(Input(shape=(STATE_SIZE,)))
        self.model.add(Dense(32, activation='relu'))
        self.model.add(Dense(32, activation='relu'))
        self.model.add(Dense(ACTION_SIZE, activation='linear'))
        self.model.compile(loss='mse', optimizer='adam')
        
    def act(self, state):
        if random.uniform(0,1) < self.epsilon:
            return random.randrange(ACTION_SIZE)
        else:
            q_values = self.model.predict(state, verbose=0)
            return np.argmax(q_values[0])
        
#     def get_scaler(self):
#         env.reset()
#         states = []
#         for _ in range(STEPS):
#             action_size = env.action_space.n
#             action = random.randrange(action_size)
#             new_state, reward, done, truncated, info = env.step(action)
#             states.append([new_state])
#             if done:
#                 break
#         scaler = StandardScaler()
#         scaler.fit(states)
#         return scaler 
    
    def remember(self, state, action, reward, new_state, done):
        self.memory.append((state, action, reward, new_state, done))
        
    def replay(self):
        if len(self.memory) < self.batch_size:
            return
        minibatch = random.sample(self.memory, self.batch_size)
        states = np.array([sample[0] for sample in minibatch])
        actions = np.array([sample[1] for sample in minibatch])
        rewards = np.array([sample[2] for sample in minibatch])
        new_states = np.array([sample[3] for sample in minibatch])
        done = np.array([sample[4] for sample in minibatch])
#         states = tf.squeeze(states, axis=1)
#         new_states = tf.squeeze(new_states, axis=1)
        target = rewards + (1 - done) * self.gamma * np.amax(self.model.predict(new_states, verbose=0), axis=1)
        target_full = self.model.predict(states, verbose=0)
        #         target_full[np.arange(self.batch_size), actions] = target
        for i, action in enumerate(actions):
            target_full[i, action] = target[i]
        self.model.fit(states, target_full, epochs=1, verbose=0)
        if self.epsilon > self.epsilon_min:
            self.epsilon *= self.epsilon_decay
```

```python
def preprocess_state(state):
    return np.identity(STATE_SIZE)[state:state+1]
```

```python
agent = DQNAgent(epsilon=EPSILON, epsilon_min=EPSILON_MIN, epsilon_decay=EPSILON_DECAY, gamma=GAMMA, batch_size=BATCH_SIZE)
for episode in range(EPISODES):
#     env.reset()
#     state = agent.scaler.transform([[0]])
    state = env.reset()[0]
    state = preprocess_state(state)
    total_reward = 0
    for step in range(STEPS):
        action = agent.act(state)
        new_state, reward, done, _, _ = env.step(action)
#         new_state = agent.scaler.transform([[new_state]])
        new_state = preprocess_state(new_state)
        break
        agent.remember(state, action, reward, new_state, done)
        state = new_state
        total_reward += reward
        if done:
            print(f"Episode {episode+1}/{EPISODES}, Reward: {total_reward}, Epsilon: {agent.epsilon}")
            break
        agent.replay()
```

```python
state = env.reset()[0]
```

```python
# state = agent.scaler.transform([[0]])
state = preprocess_state(state)
i = 0
while True:
    q_values = agent.model.predict(state, verbose=0)
    action = np.argmax(q_values[0])
    new_state, reward, done, _, _ = env.step(action)
    env.render()
    i+=1
    if done or i >=100:
        print("Done!")
        break
#     state = agent.scaler.transform([[new_state]])
    state = preprocess_state(new_state)
```
